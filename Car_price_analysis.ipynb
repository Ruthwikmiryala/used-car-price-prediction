```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from CarFeatures import clean_data
```

## Load the data

Change the file name or path if your CSV is somewhere else.

```python
df = pd.read_csv("Car details v3.csv")
print(df.shape)
df.head()
```

```python
df.info()
```

## Clean it

Mileage, engine and max_power are text like "23.4 kmpl" - they become numbers. Duplicates and rows with missing values are dropped.

```python
clean = clean_data(df)
print('rows before:', len(df), '| after:', len(clean))
clean.describe().round(1)
```

## Price is skewed, so we will predict the log of the price

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
clean['selling_price'].hist(bins=50, ax=axes[0])
axes[0].set_title('Selling price')
np.log1p(clean['selling_price']).hist(bins=50, ax=axes[1])
axes[1].set_title('log(1 + price)')
plt.show()
```

## Price vs other columns

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
axes[0].scatter(clean['year'], np.log1p(clean['selling_price']), s=5, alpha=0.3)
axes[0].set_title('log price vs year')
axes[1].scatter(np.log1p(clean['km_driven']), np.log1p(clean['selling_price']), s=5, alpha=0.3)
axes[1].set_title('log price vs log km driven')
clean.assign(log_price=np.log1p(clean['selling_price'])).boxplot(column='log_price', by='fuel', ax=axes[2])
plt.suptitle('')
plt.show()
```

```python
clean['brand'] = clean['name'].str.split().str[0]
clean.groupby('brand')['selling_price'].median().sort_values(ascending=False).head(10)
```
